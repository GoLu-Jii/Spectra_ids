```python
# ==========================================
# CODE 3 - INSPECT STATEFUL DATASETS
# ==========================================

import os
import pandas as pd
import numpy as np

BASE_DIR = "/content/drive/MyDrive/SIH_dataset"


# ==========================================
# FIND ALL STATEFUL CSV FILES
# ==========================================

stateful_files = []

for root, dirs, files in os.walk(BASE_DIR):

    for file in files:

        if (
            "stateful" in file.lower()
            and file.lower().endswith(".csv")
        ):

            stateful_files.append(
                os.path.join(root, file)
            )


# ==========================================
# INSPECT FILES
# ==========================================

file_info = []

for path in sorted(stateful_files):

    try:
        df = pd.read_csv(path)

        file_info.append({
            "folder": os.path.basename(
                os.path.dirname(
                    os.path.dirname(path)
                )
            ),
            "file": os.path.basename(path),
            "rows": len(df),
            "columns": len(df.columns),
            "column_names": list(df.columns)
        })

    except Exception as e:

        print(
            "ERROR:",
            path,
            "->",
            e
        )


# ==========================================
# DISPLAY FILE SUMMARY
# ==========================================

summary_df = pd.DataFrame(file_info)

print("=" * 70)
print("STATEFUL FILE SUMMARY")
print("=" * 70)

print(
    summary_df[
        [
            "folder",
            "file",
            "rows",
            "columns"
        ]
    ].to_string(index=False)
)


# ==========================================
# CHECK WHETHER SCHEMAS MATCH
# ==========================================

print("\n" + "=" * 70)
print("COLUMN SCHEMA CHECK")
print("=" * 70)

schema_groups = {}

for info in file_info:

    schema = tuple(
        info["column_names"]
    )

    schema_groups.setdefault(
        schema,
        []
    ).append(
        info["file"]
    )


print(
    "Number of different schemas:",
    len(schema_groups)
)

for i, (schema, files) in enumerate(
    schema_groups.items(),
    start=1
):

    print(
        f"\nSchema {i} "
        f"({len(files)} files):"
    )

    print(files)

    print("Columns:")

    for col in schema:
        print(" -", col)


# ==========================================
# CHECK DUPLICATE LIGHT FILE NAMES
# ==========================================

print("\n" + "=" * 70)
print("DUPLICATE FILE-NAME CHECK")
print("=" * 70)

filename_counts = (
    summary_df["file"]
    .value_counts()
)

print(
    filename_counts[
        filename_counts > 1
    ]
)


# ==========================================
# CHECK DUPLICATE CONTENT
# ==========================================

print("\n" + "=" * 70)
print("CONTENT DUPLICATE CHECK")
print("=" * 70)

# Only compare files with the same filename.
# We use a hash of the CSV contents.

import hashlib

hash_results = []

for path in sorted(stateful_files):

    with open(path, "rb") as f:

        file_hash = hashlib.md5(
            f.read()
        ).hexdigest()

    hash_results.append({
        "file": os.path.basename(path),
        "path": path,
        "hash": file_hash
    })


hash_df = pd.DataFrame(
    hash_results
)


duplicate_hashes = (
    hash_df[
        hash_df["hash"].duplicated(
            keep=False
        )
    ]
    .sort_values("hash")
)


if duplicate_hashes.empty:

    print(
        "No byte-identical stateful CSV files found."
    )

else:

    print(
        duplicate_hashes.to_string(
            index=False
        )
    )
```

```python
# ==========================================
# NETSPECTRA
# STATEFUL PIPELINE
# CODE 3 - CLEAN DATA LOADING
# ==========================================

import os
import pandas as pd
import numpy as np

BASE_DIR = "/content/drive/MyDrive/SIH_dataset"

SEED = 42


# ==========================================
# FILE DEFINITIONS
# ==========================================

# ------------------------------------------
# BENIGN TRAINING / DEVELOPMENT DATA
# ------------------------------------------

benign_files = [
    (
        "Benign_1/Benign",
        "stateful_features-benign_1.pcap.csv"
    ),
    (
        "Benign_1/Benign",
        "stateful_features-benign_2.pcap.csv"
    ),
    (
        "Benign_2/Benign",
        "stateful_features-_light_benign.pcap.csv"
    )
]


# ------------------------------------------
# LIGHT ATTACK DATA
# ------------------------------------------

light_attack_files = [
    (
        "Attacks/Attacks",
        "stateful_features-light_image.pcap.csv"
    ),
    (
        "Attacks/Attacks",
        "stateful_features-light_audio.pcap.csv"
    ),
    (
        "Attacks/Attacks",
        "stateful_features-light_compressed.pcap.csv"
    ),
    (
        "Attacks/Attacks",
        "stateful_features-light_video.pcap.csv"
    ),
    (
        "Attacks/Attacks",
        "stateful_features-light_exe.pcap.csv"
    ),
    (
        "Attacks/Attacks",
        "stateful_features-light_text.pcap.csv"
    )
]


# ==========================================
# LOAD FUNCTION
# ==========================================

def load_stateful_file(
    relative_folder,
    filename,
    label
):

    file_path = os.path.join(
        BASE_DIR,
        relative_folder,
        filename
    )

    if not os.path.exists(file_path):

        raise FileNotFoundError(
            f"File not found:\n{file_path}"
        )

    df = pd.read_csv(file_path)

    # Target
    df["label"] = label

    # Original capture
    df["source"] = filename

    # Full path for traceability
    df["source_path"] = file_path

    return df


# ==========================================
# LOAD BENIGN
# ==========================================

benign_dfs = []

print("=" * 70)
print("LOADING BENIGN STATEFUL DATA")
print("=" * 70)

for folder, filename in benign_files:

    df = load_stateful_file(
        folder,
        filename,
        label=0
    )

    benign_dfs.append(df)

    print(
        f"{filename:55s} -> {df.shape}"
    )


# ==========================================
# LOAD LIGHT ATTACKS
# ==========================================

attack_dfs = []

print("\n" + "=" * 70)
print("LOADING LIGHT ATTACK STATEFUL DATA")
print("=" * 70)

for folder, filename in light_attack_files:

    df = load_stateful_file(
        folder,
        filename,
        label=1
    )

    attack_dfs.append(df)

    print(
        f"{filename:55s} -> {df.shape}"
    )


# ==========================================
# COMBINE
# ==========================================

df_stateful = pd.concat(
    benign_dfs + attack_dfs,
    ignore_index=True
)


# ==========================================
# FINAL SUMMARY
# ==========================================

print("\n" + "=" * 70)
print("FINAL STATEFUL DATASET")
print("=" * 70)

print(
    "Shape:",
    df_stateful.shape
)


print("\nClass distribution:")

print(
    df_stateful["label"]
    .value_counts()
    .sort_index()
)


print("\nClass percentages:")

print(
    df_stateful["label"]
    .value_counts(
        normalize=True
    )
    .sort_index()
    .mul(100)
    .round(2)
)


print("\nSource distribution:")

print(
    df_stateful["source"]
    .value_counts()
)


# ==========================================
# VERIFY NO DUPLICATE SOURCE FILES
# ==========================================

print("\n" + "=" * 70)
print("SOURCE FILE CHECK")
print("=" * 70)

print(
    "Unique source files:",
    df_stateful["source"].nunique()
)

print(
    df_stateful[
        ["source", "label"]
    ]
    .drop_duplicates()
    .sort_values(
        ["label", "source"]
    )
    .to_string(index=False)
)


# ==========================================
# VERIFY DUPLICATE ROWS
# ==========================================

print("\n" + "=" * 70)
print("DUPLICATE ROW CHECK")
print("=" * 70)

duplicate_count = (
    df_stateful
    .drop(columns=["source_path"])
    .duplicated()
    .sum()
)

print(
    "Duplicate rows:",
    duplicate_count
)

print(
    "Duplicate percentage:",
    round(
        duplicate_count
        / len(df_stateful)
        * 100,
        4
    ),
    "%"
)
```

```python
 # ==========================================
# CODE 6 - STATEFUL FEATURE ENGINEERING
# NO PLOTS
# ==========================================

import ast
import re
import pandas as pd
import numpy as np


# ==========================================
# START WITH COPY
# ==========================================

df_features = df_stateful.copy()


# ==========================================
# 1. REMOVE CONSTANT FEATURES
# ==========================================

constant_features = [
    "NS_frequency",
    "CNAME_frequency",
    "SOA_frequency",
    "NULL_frequency",
    "HINFO_frequency",
    "MX_frequency",
    "SRV_frequency",
    "OPT_frequency",
    "a_records"
]


df_features = df_features.drop(
    columns=constant_features
)


print("=" * 70)
print("REMOVED CONSTANT FEATURES")
print("=" * 70)

for feature in constant_features:
    print("-", feature)


# ==========================================
# 2. REMOVE EXACT DUPLICATE INFORMATION
# ==========================================

# rr and A_frequency have identical values
# in this dataset.

print("\n" + "=" * 70)
print("CHECK rr vs A_frequency")
print("=" * 70)

print(
    "Identical:",
    df_features["rr"].equals(
        df_features["A_frequency"]
    )
)


# Keep rr and remove A_frequency
df_features = df_features.drop(
    columns=["A_frequency"]
)


# ==========================================
# 3. REMOVE USELESS OBJECT COLUMNS
# ==========================================

# distinct_ip is set() for every row.
# source/source_path are metadata, not features.

useless_columns = [
    "distinct_ip",
    "source",
    "source_path"
]

df_features = df_features.drop(
    columns=useless_columns
)


# ==========================================
# 4. FUNCTION TO COUNT SET-LIKE VALUES
# ==========================================

def count_set_values(value):

    if pd.isna(value):
        return 0

    value = str(value).strip()

    # Empty set
    if value in [
        "set()",
        "{}",
        "None",
        "nan"
    ]:
        return 0

    try:

        parsed = ast.literal_eval(value)

        if isinstance(parsed, (set, list, tuple)):
            return len(parsed)

        if isinstance(parsed, dict):
            return len(parsed)

    except Exception:
        pass

    # Fallback
    return 1


# ==========================================
# 5. COUNTRY FEATURES
# ==========================================

df_features["country_count"] = (
    df_features["unique_country"]
    .apply(count_set_values)
)


# ==========================================
# 6. ASN FEATURES
# ==========================================

df_features["asn_count"] = (
    df_features["unique_asn"]
    .apply(count_set_values)
)


# ==========================================
# 7. DOMAIN FEATURES
# ==========================================

df_features["domain_count"] = (
    df_features["distinct_domains"]
    .apply(count_set_values)
)


# ==========================================
# 8. TTL FEATURES
# ==========================================

def parse_ttl_values(value):

    if pd.isna(value):
        return []

    value = str(value).strip()

    try:

        parsed = ast.literal_eval(value)

        if isinstance(parsed, (list, tuple)):
            return [
                float(x)
                for x in parsed
                if str(x).strip()
            ]

    except Exception:
        pass

    # Fallback: find numbers
    numbers = re.findall(
        r"\d+(?:\.\d+)?",
        value
    )

    return [
        float(x)
        for x in numbers
    ]


ttl_values = (
    df_features["unique_ttl"]
    .apply(parse_ttl_values)
)


df_features["ttl_unique_count"] = (
    ttl_values.apply(len)
)


df_features["ttl_min"] = (
    ttl_values.apply(
        lambda x: min(x)
        if len(x) > 0 else 0
    )
)


df_features["ttl_max"] = (
    ttl_values.apply(
        lambda x: max(x)
        if len(x) > 0 else 0
    )
)


df_features["ttl_range"] = (
    df_features["ttl_max"]
    - df_features["ttl_min"]
)


# ==========================================
# 9. REVERSE DNS FEATURES
# ==========================================

df_features["reverse_dns_known"] = (
    (
        df_features["reverse_dns"]
        .astype(str)
        .str.lower()
        != "unknown"
    )
    .astype(int)
)


# ==========================================
# 10. RESOURCE RECORD FEATURES
# ==========================================

df_features["is_A_record"] = (
    df_features["rr_type"]
    .astype(str)
    .str.contains(
        "A",
        regex=False
    )
    .astype(int)
)


df_features["is_PTR_record"] = (
    df_features["rr_type"]
    .astype(str)
    .str.contains(
        "PTR",
        regex=False
    )
    .astype(int)
)


df_features["is_TXT_record"] = (
    df_features["rr_type"]
    .astype(str)
    .str.contains(
        "TXT",
        regex=False
    )
    .astype(int)
)


df_features["rr_type_count"] = (
    df_features["rr_type"]
    .apply(count_set_values)
)


# ==========================================
# 11. BEHAVIORAL RATIOS
# ==========================================

safe_rr_count = (
    df_features["rr_count"]
    .clip(lower=1)
)

safe_ttl_mean = (
    df_features["ttl_mean"]
    .clip(lower=1)
)


df_features["ptr_per_rr"] = (
    df_features["PTR_frequency"]
    / safe_rr_count
)


df_features["rr_entropy_per_length"] = (
    df_features["rr_name_entropy"]
    / df_features["rr_name_length"].clip(
        lower=1
    )
)


df_features["ttl_variance_ratio"] = (
    df_features["ttl_variance"]
    / safe_ttl_mean
)


# ==========================================
# 12. DROP RAW OBJECT COLUMNS
# ==========================================

object_columns_to_drop = [
    "rr_type",
    "unique_country",
    "unique_asn",
    "distinct_domains",
    "reverse_dns",
    "unique_ttl"
]


df_features = df_features.drop(
    columns=object_columns_to_drop
)


# ==========================================
# 13. CHECK FINAL COLUMNS
# ==========================================

print("\n" + "=" * 70)
print("FINAL FEATURE COLUMNS AFTER ENGINEERING")
print("=" * 70)

for i, column in enumerate(
    df_features.columns,
    start=1
):

    print(
        f"{i:2d}. {column} "
        f"({df_features[column].dtype})"
    )


# ==========================================
# 14. CHECK FOR NON-NUMERIC COLUMNS
# ==========================================

non_numeric = (
    df_features
    .select_dtypes(
        exclude=[np.number]
    )
    .columns
    .tolist()
)


print("\n" + "=" * 70)
print("NON-NUMERIC COLUMNS")
print("=" * 70)

print(non_numeric)


# ==========================================
# 15. MISSING VALUES
# ==========================================

print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

print(
    df_features
    .isnull()
    .sum()
    .sort_values(
        ascending=False
    )
)


# ==========================================
# 16. FEATURE COUNT
# ==========================================

model_features = [
    col
    for col in df_features.columns
    if col != "label"
]


print("\n" + "=" * 70)
print("MODEL FEATURE COUNT")
print("=" * 70)

print(
    "Number of model features:",
    len(model_features)
)


# ==========================================
# 17. DESCRIPTIVE STATISTICS
# ==========================================

print("\n" + "=" * 70)
print("ENGINEERED FEATURE STATISTICS")
print("=" * 70)

print(
    df_features[
        model_features
    ]
    .describe()
    .T
    .round(4)
)
```

```python
import ast
import pandas as pd
import numpy as np

print("=" * 70)
print("1. CHECK rr vs A_frequency")
print("=" * 70)

rr_values = df_stateful["rr"]
a_values = df_stateful["A_frequency"]

mismatch_mask = ~np.isclose(
    rr_values.astype(float),
    a_values.astype(float),
    equal_nan=True
)

mismatch_count = mismatch_mask.sum()

print("Total rows:", len(df_stateful))
print("Mismatched rows:", mismatch_count)
print("Mismatch percentage:", round(mismatch_count / len(df_stateful) * 100, 4), "%")

if mismatch_count > 0:
    print("\nMaximum absolute difference:",
          np.max(np.abs(
              rr_values[mismatch_mask].astype(float)
              - a_values[mismatch_mask].astype(float)
          )))

    print("\nSample mismatches:")
    print(
        df_stateful.loc[
            mismatch_mask,
            ["rr", "A_frequency", "rr_type", "rr_count", "label"]
        ].head(20)
    )
else:
    print("rr and A_frequency are identical.")


print("\n" + "=" * 70)
print("2. EXACT rr_type PARSING")
print("=" * 70)

def parse_rr_type(value):
    """
    Convert rr_type into a Python set safely.
    Examples:
        "{'A'}"              -> {'A'}
        "{'AAAA', 'A'}"      -> {'AAAA', 'A'}
        "{'PTR'}"            -> {'PTR'}
        "set()"              -> set()
        None                 -> set()
    """

    if pd.isna(value):
        return set()

    if isinstance(value, set):
        return value

    try:
        parsed = ast.literal_eval(str(value))

        if isinstance(parsed, set):
            return parsed

        if isinstance(parsed, (list, tuple)):
            return set(parsed)

        return set()

    except Exception:
        return set()


parsed_rr_types = df_stateful["rr_type"].apply(parse_rr_type)

print("Unique parsed RR-type combinations:")
print(parsed_rr_types.value_counts().head(20))


print("\n" + "=" * 70)
print("3. REBUILD RR-TYPE FEATURES")
print("=" * 70)

# Exact matching instead of substring matching
df_features["is_A_record"] = parsed_rr_types.apply(
    lambda x: int("A" in x)
)

df_features["is_PTR_record"] = parsed_rr_types.apply(
    lambda x: int("PTR" in x)
)

df_features["is_TXT_record"] = parsed_rr_types.apply(
    lambda x: int("TXT" in x)
)

df_features["is_AAAA_record"] = parsed_rr_types.apply(
    lambda x: int("AAAA" in x)
)

df_features["rr_type_count"] = parsed_rr_types.apply(len)


print("A records:", df_features["is_A_record"].sum())
print("AAAA records:", df_features["is_AAAA_record"].sum())
print("PTR records:", df_features["is_PTR_record"].sum())
print("TXT records:", df_features["is_TXT_record"].sum())


print("\n" + "=" * 70)
print("4. RESTORE A_frequency FOR NOW")
print("=" * 70)

# Since rr != A_frequency, keep both until we analyze their predictive value.
df_features["A_frequency"] = df_stateful["A_frequency"].values

print("A_frequency added.")
print("A_frequency unique values:",
      df_features["A_frequency"].nunique())

print("\nCorrelation between rr and A_frequency:")
print(
    round(
        df_features[["rr", "A_frequency"]].corr().iloc[0, 1],
        6
    )
)


print("\n" + "=" * 70)
print("5. FINAL NUMERIC FEATURE VALIDATION")
print("=" * 70)

# Remove columns that must never be used as model features
drop_for_model = [
    "label"
]

X_temp = df_features.drop(columns=drop_for_model)

# Make sure everything is numeric
non_numeric = X_temp.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Non-numeric columns:", non_numeric)

print("\nMissing values:")
print(X_temp.isna().sum())

print("\nNumber of features:", X_temp.shape[1])

print("\nFinal feature list:")
for i, col in enumerate(X_temp.columns, 1):
    print(f"{i:2d}. {col}")

print("\n" + "=" * 70)
print("FEATURE ENGINEERING CHECK COMPLETE")
print("=" * 70)
```

```python
from sklearn.model_selection import StratifiedGroupKFold
import numpy as np
import pandas as pd

print("=" * 70)
print("CORRECTED STATEFUL TRAIN / VALIDATION / TEST SPLIT")
print("=" * 70)

# ------------------------------------------------------------
# 1. Prepare data
# ------------------------------------------------------------

X = df_features.drop(columns=["label"]).copy()
y = df_features["label"].copy()

# Source/capture is used ONLY for grouping.
# It will NOT be given to the model.
groups = df_stateful["source"].copy()

print("Total samples :", len(X))
print("Total features:", X.shape[1])
print("Total groups  :", groups.nunique())


# ------------------------------------------------------------
# 2. OUTER SPLIT
#
# Create 3 group-level folds.
# One complete fold will become the final TEST set.
# ------------------------------------------------------------

outer_cv = StratifiedGroupKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

outer_folds = list(
    outer_cv.split(X, y, groups)
)

print("\n" + "=" * 70)
print("OUTER FOLDS")
print("=" * 70)

for i, (train_idx_outer, test_idx_outer) in enumerate(outer_folds):

    test_sources = sorted(
        groups.iloc[test_idx_outer].unique()
    )

    test_labels = y.iloc[test_idx_outer]

    print(f"\nFOLD {i}")
    print("-" * 70)

    print("Test rows:", len(test_idx_outer))
    print("Test sources:")
    for source in test_sources:
        print("  -", source)

    print("\nClass distribution:")
    print(test_labels.value_counts().sort_index().to_dict())


# ------------------------------------------------------------
# 3. Select OUTER FOLD 1 as final TEST
#
# We keep this fixed so the experiment is reproducible.
# ------------------------------------------------------------

_, test_idx = outer_folds[1]

test_sources = set(
    groups.iloc[test_idx].unique()
)

remaining_idx = np.setdiff1d(
    np.arange(len(X)),
    test_idx
)

print("\n" + "=" * 70)
print("FINAL TEST SET")
print("=" * 70)

print("Test rows:", len(test_idx))
print("Test sources:")
for source in sorted(test_sources):
    print("  -", source)

print("\nTest class distribution:")
print(
    y.iloc[test_idx]
    .value_counts()
    .sort_index()
)


# ------------------------------------------------------------
# 4. INNER SPLIT
#
# Split the remaining sources into TRAIN and VALIDATION.
# ------------------------------------------------------------

X_remaining = X.iloc[remaining_idx]
y_remaining = y.iloc[remaining_idx]
groups_remaining = groups.iloc[remaining_idx]

inner_cv = StratifiedGroupKFold(
    n_splits=2,
    shuffle=True,
    random_state=42
)

inner_folds = list(
    inner_cv.split(
        X_remaining,
        y_remaining,
        groups_remaining
    )
)

# Use inner fold 0:
# inner train -> TRAIN
# inner test  -> VALIDATION

train_relative_idx, val_relative_idx = inner_folds[0]

train_idx = remaining_idx[train_relative_idx]
val_idx = remaining_idx[val_relative_idx]


# ------------------------------------------------------------
# 5. Create final datasets
# ------------------------------------------------------------

X_train = X.iloc[train_idx].copy()
y_train = y.iloc[train_idx].copy()

X_val = X.iloc[val_idx].copy()
y_val = y.iloc[val_idx].copy()

X_test = X.iloc[test_idx].copy()
y_test = y.iloc[test_idx].copy()


# ------------------------------------------------------------
# 6. Get source groups
# ------------------------------------------------------------

train_sources = set(
    groups.iloc[train_idx].unique()
)

val_sources = set(
    groups.iloc[val_idx].unique()
)

test_sources = set(
    groups.iloc[test_idx].unique()
)


# ------------------------------------------------------------
# 7. Display final split
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL SPLIT")
print("=" * 70)

print("\nTRAIN")
print("-" * 70)
print("Rows:", len(X_train))
print("Sources:")
for source in sorted(train_sources):
    print("  -", source)

print("\nClass distribution:")
print(y_train.value_counts().sort_index())


print("\nVALIDATION")
print("-" * 70)
print("Rows:", len(X_val))
print("Sources:")
for source in sorted(val_sources):
    print("  -", source)

print("\nClass distribution:")
print(y_val.value_counts().sort_index())


print("\nTEST")
print("-" * 70)
print("Rows:", len(X_test))
print("Sources:")
for source in sorted(test_sources):
    print("  -", source)

print("\nClass distribution:")
print(y_test.value_counts().sort_index())


# ------------------------------------------------------------
# 8. LEAKAGE CHECK
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL SOURCE LEAKAGE CHECK")
print("=" * 70)

train_val_overlap = train_sources.intersection(val_sources)
train_test_overlap = train_sources.intersection(test_sources)
val_test_overlap = val_sources.intersection(test_sources)

print("Train ∩ Validation:", train_val_overlap)
print("Train ∩ Test      :", train_test_overlap)
print("Validation ∩ Test :", val_test_overlap)


# ------------------------------------------------------------
# 9. Row overlap check
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ROW INDEX OVERLAP CHECK")
print("=" * 70)

train_set = set(train_idx)
val_set = set(val_idx)
test_set = set(test_idx)

print("Train ∩ Validation:", len(train_set.intersection(val_set)))
print("Train ∩ Test      :", len(train_set.intersection(test_set)))
print("Validation ∩ Test :", len(val_set.intersection(test_set)))


# ------------------------------------------------------------
# 10. Class percentage
# ------------------------------------------------------------

def show_distribution(name, target):

    counts = target.value_counts().sort_index()

    percentages = (
        target.value_counts(normalize=True)
        .sort_index()
        .mul(100)
        .round(2)
    )

    print(f"\n{name}")
    print("Benign (0):", counts.get(0, 0),
          f"({percentages.get(0, 0)}%)")

    print("Attack (1):", counts.get(1, 0),
          f"({percentages.get(1, 0)}%)")


print("\n" + "=" * 70)
print("CLASS DISTRIBUTION")
print("=" * 70)

show_distribution("TRAIN", y_train)
show_distribution("VALIDATION", y_val)
show_distribution("TEST", y_test)


# ------------------------------------------------------------
# 11. Final shapes
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL SHAPES")
print("=" * 70)

print("X_train:", X_train.shape)
print("X_val  :", X_val.shape)
print("X_test :", X_test.shape)

print("y_train:", y_train.shape)
print("y_val  :", y_val.shape)
print("y_test :", y_test.shape)


# ------------------------------------------------------------
# 12. Safety assertions
# ------------------------------------------------------------

assert len(train_val_overlap) == 0
assert len(train_test_overlap) == 0
assert len(val_test_overlap) == 0

assert len(train_set.intersection(val_set)) == 0
assert len(train_set.intersection(test_set)) == 0
assert len(val_set.intersection(test_set)) == 0

assert set(y_train.unique()) == {0, 1}
assert set(y_val.unique()) == {0, 1}
assert set(y_test.unique()) == {0, 1}

print("\n" + "=" * 70)
print("✓ SPLIT PASSED ALL LEAKAGE CHECKS")
print("=" * 70)
```

```python
from sklearn.metrics import roc_auc_score
from sklearn.feature_selection import mutual_info_classif
import pandas as pd
import numpy as np

print("=" * 80)
print("TRAINING-ONLY FEATURE ANALYSIS")
print("=" * 80)

# ============================================================
# 1. Make sure we are using TRAIN ONLY
# ============================================================

X_train_analysis = X_train.copy()
y_train_analysis = y_train.copy()

print("Training samples :", X_train_analysis.shape[0])
print("Training features:", X_train_analysis.shape[1])

assert len(X_train_analysis) == len(y_train_analysis)

print("\nValidation and test data are NOT used in this analysis.")


# ============================================================
# 2. Class-wise feature means
# ============================================================

print("\n" + "=" * 80)
print("CLASS-WISE FEATURE MEANS")
print("=" * 80)

class_means = pd.DataFrame({
    "Benign_Mean": X_train_analysis[y_train_analysis == 0].mean(),
    "Attack_Mean": X_train_analysis[y_train_analysis == 1].mean()
})

class_means["Absolute_Difference"] = (
    class_means["Attack_Mean"]
    - class_means["Benign_Mean"]
).abs()

class_means = class_means.sort_values(
    "Absolute_Difference",
    ascending=False
)

print(class_means.round(4))


# ============================================================
# 3. Univariate ROC-AUC
# ============================================================

print("\n" + "=" * 80)
print("UNIVARIATE ROC-AUC")
print("=" * 80)

auc_results = []

for feature in X_train_analysis.columns:

    values = X_train_analysis[feature]

    # AUC requires more than one unique value
    if values.nunique() <= 1:
        auc = 0.5
        direction = "constant"

    else:
        try:
            auc_raw = roc_auc_score(
                y_train_analysis,
                values
            )

            # AUC below 0.5 simply means the reverse
            # direction is more informative.
            auc = max(auc_raw, 1 - auc_raw)

            if auc_raw >= 0.5:
                direction = "higher_attack"
            else:
                direction = "lower_attack"

        except Exception:
            auc = 0.5
            direction = "error"

    auc_results.append({
        "feature": feature,
        "roc_auc": auc,
        "direction": direction
    })

auc_df = pd.DataFrame(auc_results)

auc_df = auc_df.sort_values(
    "roc_auc",
    ascending=False
).reset_index(drop=True)

print(
    auc_df.to_string(
        index=False,
        formatters={
            "roc_auc": "{:.4f}".format
        }
    )
)


# ============================================================
# 4. Mutual Information
# ============================================================

print("\n" + "=" * 80)
print("MUTUAL INFORMATION")
print("=" * 80)

# Mutual information can be computationally expensive.
# 25k rows × 27 features is still manageable.

mi_values = mutual_info_classif(
    X_train_analysis,
    y_train_analysis,
    random_state=42
)

mi_df = pd.DataFrame({
    "feature": X_train_analysis.columns,
    "mutual_information": mi_values
})

mi_df = mi_df.sort_values(
    "mutual_information",
    ascending=False
).reset_index(drop=True)

print(
    mi_df.to_string(
        index=False,
        formatters={
            "mutual_information": "{:.6f}".format
        }
    )
)


# ============================================================
# 5. Correlation between features
# ============================================================

print("\n" + "=" * 80)
print("HIGH FEATURE CORRELATIONS")
print("=" * 80)

corr_matrix = X_train_analysis.corr()

correlation_pairs = []

features = corr_matrix.columns

for i in range(len(features)):
    for j in range(i + 1, len(features)):

        corr_value = corr_matrix.iloc[i, j]

        if pd.isna(corr_value):
            continue

        correlation_pairs.append({
            "feature_1": features[i],
            "feature_2": features[j],
            "correlation": abs(corr_value)
        })

corr_df = pd.DataFrame(correlation_pairs)

corr_df = corr_df.sort_values(
    "correlation",
    ascending=False
).reset_index(drop=True)

print(
    corr_df.head(30).to_string(
        index=False,
        formatters={
            "correlation": "{:.4f}".format
        }
    )
)


# ============================================================
# 6. Combine the rankings
# ============================================================

print("\n" + "=" * 80)
print("COMBINED FEATURE RANKING")
print("=" * 80)

feature_ranking = auc_df.merge(
    mi_df,
    on="feature"
)

feature_ranking = feature_ranking.merge(
    class_means.reset_index().rename(
        columns={"index": "feature"}
    ),
    on="feature"
)

# Rank by ROC-AUC
feature_ranking["AUC_Rank"] = (
    feature_ranking["roc_auc"]
    .rank(
        ascending=False,
        method="min"
    )
)

# Rank by Mutual Information
feature_ranking["MI_Rank"] = (
    feature_ranking["mutual_information"]
    .rank(
        ascending=False,
        method="min"
    )
)

# Average rank
feature_ranking["Average_Rank"] = (
    feature_ranking["AUC_Rank"]
    + feature_ranking["MI_Rank"]
) / 2

feature_ranking = feature_ranking.sort_values(
    "Average_Rank"
).reset_index(drop=True)

print(
    feature_ranking[
        [
            "feature",
            "roc_auc",
            "mutual_information",
            "AUC_Rank",
            "MI_Rank",
            "Average_Rank"
        ]
    ].to_string(
        index=False,
        formatters={
            "roc_auc": "{:.4f}".format,
            "mutual_information": "{:.6f}".format,
            "AUC_Rank": "{:.0f}".format,
            "MI_Rank": "{:.0f}".format,
            "Average_Rank": "{:.1f}".format
        }
    )
)


# ============================================================
# 7. Most useful features
# ============================================================

print("\n" + "=" * 80)
print("TOP FEATURES")
print("=" * 80)

top_features = feature_ranking.head(15)["feature"].tolist()

for i, feature in enumerate(top_features, 1):
    print(f"{i:2d}. {feature}")


# ============================================================
# 8. Final checks
# ============================================================

print("\n" + "=" * 80)
print("ANALYSIS CHECKS")
print("=" * 80)

print("NaN in AUC results:",
      auc_df.isna().sum().sum())

print("NaN in MI results:",
      mi_df.isna().sum().sum())

print("NaN in feature ranking:",
      feature_ranking.isna().sum().sum())

print("\n" + "=" * 80)
print("✓ TRAINING-ONLY FEATURE ANALYSIS COMPLETE")
print("=" * 80)
```

```python
import xgboost as xgb
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

print("=" * 80)
print("XGBOOST BASELINE MODEL")
print("=" * 80)

print("XGBoost version:", xgb.__version__)


# ============================================================
# 1. SELECT MODEL FEATURES
# ============================================================

model_features = [
    "rr",
    "PTR_frequency",
    "rr_count",
    "rr_name_entropy",
    "rr_name_length",
    "distinct_ns",
    "ttl_mean",
    "ttl_variance",
    "country_count",
    "asn_count",
    "ttl_unique_count",
    "ttl_min",
    "ttl_max",
    "ttl_range",
    "reverse_dns_known",
    "is_A_record",
    "is_PTR_record",
    "rr_entropy_per_length",
    "ttl_variance_ratio"
]

print("\nNumber of selected features:", len(model_features))

print("\nSelected features:")
for i, feature in enumerate(model_features, 1):
    print(f"{i:2d}. {feature}")


# ============================================================
# 2. CREATE FINAL MODEL MATRICES
# ============================================================

X_train_model = X_train[model_features].copy()
X_val_model = X_val[model_features].copy()
X_test_model = X_test[model_features].copy()

print("\n" + "=" * 80)
print("MODEL MATRIX SHAPES")
print("=" * 80)

print("X_train:", X_train_model.shape)
print("X_val  :", X_val_model.shape)
print("X_test :", X_test_model.shape)


# ============================================================
# 3. CHECK DATA
# ============================================================

assert X_train_model.isna().sum().sum() == 0
assert X_val_model.isna().sum().sum() == 0
assert X_test_model.isna().sum().sum() == 0

assert list(X_train_model.columns) == model_features
assert list(X_val_model.columns) == model_features
assert list(X_test_model.columns) == model_features


# ============================================================
# 4. CLASS IMBALANCE
# ============================================================

negative = (y_train == 0).sum()
positive = (y_train == 1).sum()

scale_pos_weight = negative / positive

print("\n" + "=" * 80)
print("CLASS IMBALANCE")
print("=" * 80)

print("Benign samples:", negative)
print("Attack samples:", positive)
print(
    "scale_pos_weight:",
    round(scale_pos_weight, 4)
)


# ============================================================
# 5. CREATE XGBOOST MODEL
# ============================================================

model = xgb.XGBClassifier(
    n_estimators=1000,
    learning_rate=0.05,
    max_depth=6,
    min_child_weight=2,
    subsample=0.90,
    colsample_bytree=0.90,

    # Handle class imbalance
    scale_pos_weight=scale_pos_weight,

    # Regularization
    reg_alpha=0.1,
    reg_lambda=2.0,

    objective="binary:logistic",
    eval_metric=["auc", "aucpr", "logloss"],

    random_state=42,
    n_jobs=-1,
    tree_method="hist"
)


# ============================================================
# 6. TRAIN
# ============================================================

print("\n" + "=" * 80)
print("TRAINING XGBOOST")
print("=" * 80)

try:

    model.fit(
        X_train_model,
        y_train,
        eval_set=[
            (X_train_model, y_train),
            (X_val_model, y_val)
        ],
        verbose=False,
        early_stopping_rounds=50
    )

    print("Training completed with early stopping.")

except TypeError:

    print(
        "Early stopping argument not supported by this "
        "XGBoost version. Training without early stopping."
    )

    model.fit(
        X_train_model,
        y_train,
        eval_set=[
            (X_train_model, y_train),
            (X_val_model, y_val)
        ],
        verbose=False
    )


# ============================================================
# 7. PREDICTIONS
# ============================================================

train_prob = model.predict_proba(
    X_train_model
)[:, 1]

val_prob = model.predict_proba(
    X_val_model
)[:, 1]

test_prob = model.predict_proba(
    X_test_model
)[:, 1]


# ============================================================
# 8. METRIC FUNCTION
# ============================================================

def evaluate_model(y_true, probabilities, threshold=0.5):

    predictions = (
        probabilities >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        predictions
    ).ravel()

    accuracy = accuracy_score(
        y_true,
        predictions
    )

    precision = precision_score(
        y_true,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        predictions,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_true,
        probabilities
    )

    pr_auc = average_precision_score(
        y_true,
        probabilities
    )

    fpr = fp / (fp + tn)

    return {
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "ROC-AUC": roc_auc,
        "PR-AUC": pr_auc,
        "FPR": fpr,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }


# ============================================================
# 9. TRAINING PERFORMANCE
# ============================================================

train_metrics = evaluate_model(
    y_train,
    train_prob,
    threshold=0.5
)

print("\n" + "=" * 80)
print("TRAINING METRICS @ THRESHOLD 0.50")
print("=" * 80)

for metric, value in train_metrics.items():

    if metric in ["TN", "FP", "FN", "TP"]:
        print(f"{metric:10s}: {value}")

    else:
        print(f"{metric:10s}: {value:.4f}")


# ============================================================
# 10. VALIDATION PERFORMANCE
# ============================================================

val_metrics = evaluate_model(
    y_val,
    val_prob,
    threshold=0.5
)

print("\n" + "=" * 80)
print("VALIDATION METRICS @ THRESHOLD 0.50")
print("=" * 80)

for metric, value in val_metrics.items():

    if metric in ["TN", "FP", "FN", "TP"]:
        print(f"{metric:10s}: {value}")

    else:
        print(f"{metric:10s}: {value:.4f}")


# ============================================================
# 11. FIND BEST VALIDATION F1 THRESHOLD
#
# This threshold is selected ONLY on validation data.
# The test set is not used.
# ============================================================

thresholds = np.arange(
    0.05,
    0.96,
    0.01
)

threshold_results = []

for threshold in thresholds:

    predictions = (
        val_prob >= threshold
    ).astype(int)

    precision = precision_score(
        y_val,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_val,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_val,
        predictions,
        zero_division=0
    )

    threshold_results.append({
        "threshold": threshold,
        "precision": precision,
        "recall": recall,
        "f1": f1
    })

threshold_df = pd.DataFrame(
    threshold_results
)

best_row = threshold_df.loc[
    threshold_df["f1"].idxmax()
]

best_threshold = float(
    best_row["threshold"]
)


# ============================================================
# 12. VALIDATION METRICS AT OPTIMAL THRESHOLD
# ============================================================

val_tuned_metrics = evaluate_model(
    y_val,
    val_prob,
    threshold=best_threshold
)

print("\n" + "=" * 80)
print("BEST VALIDATION THRESHOLD")
print("=" * 80)

print("Threshold:",
      round(best_threshold, 2))

print("\nValidation metrics at selected threshold:")

for metric, value in val_tuned_metrics.items():

    if metric in ["TN", "FP", "FN", "TP"]:
        print(f"{metric:10s}: {value}")

    else:
        print(f"{metric:10s}: {value:.4f}")


# ============================================================
# 13. MODEL INFORMATION
# ============================================================

print("\n" + "=" * 80)
print("MODEL INFORMATION")
print("=" * 80)

print("Best iteration:",
      getattr(model, "best_iteration", "N/A"))

print(
    "Best validation score:",
    getattr(model, "best_score", "N/A")
)

print("\nModel training complete.")
print("IMPORTANT: Test data has NOT been used for tuning.")
```

```python
import ast
import pandas as pd
import numpy as np

print("=" * 80)
print("RAW STATEFUL FEATURE INFORMATION-LOSS AUDIT")
print("=" * 80)


# ============================================================
# 1. Original stateful feature columns
# ============================================================

raw_feature_cols = [
    "rr",
    "A_frequency",
    "NS_frequency",
    "CNAME_frequency",
    "SOA_frequency",
    "NULL_frequency",
    "PTR_frequency",
    "HINFO_frequency",
    "MX_frequency",
    "TXT_frequency",
    "AAAA_frequency",
    "SRV_frequency",
    "OPT_frequency",
    "rr_type",
    "rr_count",
    "rr_name_entropy",
    "rr_name_length",
    "distinct_ns",
    "distinct_ip",
    "unique_country",
    "unique_asn",
    "distinct_domains",
    "reverse_dns",
    "a_records",
    "unique_ttl",
    "ttl_mean",
    "ttl_variance"
]

print("Original stateful features:", len(raw_feature_cols))
print("Current model features   :", len(model_features))


# ============================================================
# 2. Canonicalize values
#
# We need deterministic representations for sets/dicts/lists.
# ============================================================

def canonicalize(value):

    # Missing / None
    if value is None:
        return "__NONE__"

    try:
        if pd.isna(value):
            return "__NONE__"
    except Exception:
        pass

    # Already a Python set
    if isinstance(value, set):
        items = [
            canonicalize(v)
            for v in value
        ]

        return (
            "SET",
            tuple(sorted(items, key=str))
        )

    # Dictionary
    if isinstance(value, dict):
        items = []

        for k, v in value.items():
            items.append(
                (
                    canonicalize(k),
                    canonicalize(v)
                )
            )

        return (
            "DICT",
            tuple(
                sorted(
                    items,
                    key=str
                )
            )
        )

    # List / tuple
    if isinstance(value, (list, tuple)):
        return (
            "LIST",
            tuple(
                canonicalize(v)
                for v in value
            )
        )

    # Strings that represent Python structures
    if isinstance(value, str):

        text = value.strip()

        # Try parsing only structure-like strings
        if (
            text.startswith(("{", "[", "("))
            or text in ["set()", "None"]
        ):

            try:
                parsed = ast.literal_eval(text)

                if parsed is None:
                    return "__NONE__"

                return canonicalize(parsed)

            except Exception:
                pass

        return text

    # Numeric values
    if isinstance(value, (int, float, np.integer, np.floating)):

        if isinstance(value, float) and np.isnan(value):
            return "__NONE__"

        return float(value)

    return str(value)


# ============================================================
# 3. Create complete RAW feature keys
# ============================================================

def make_raw_key(dataframe):

    return dataframe[
        raw_feature_cols
    ].apply(
        lambda row: tuple(
            canonicalize(v)
            for v in row
        ),
        axis=1
    )


print("\nCreating raw feature keys...")

train_raw_keys = make_raw_key(
    df_stateful.loc[X_train.index]
)

val_raw_keys = make_raw_key(
    df_stateful.loc[X_val.index]
)

print("Raw training keys created.")
print("Raw validation keys created.")


# ============================================================
# 4. Create engineered keys for comparison
# ============================================================

train_engineered_keys = make_feature_key(
    X_train
)

val_engineered_keys = make_feature_key(
    X_val
)


# ============================================================
# 5. Training RAW pattern lookup
# ============================================================

train_raw_pattern_df = pd.DataFrame({
    "key": train_raw_keys.values,
    "label": y_train.values
})

train_raw_stats = (
    train_raw_pattern_df
    .groupby("key")["label"]
    .agg(
        train_count="size",
        train_attack_count="sum",
        train_attack_rate="mean"
    )
    .reset_index()
)


# ============================================================
# 6. Validation analysis
# ============================================================

raw_analysis = pd.DataFrame({
    "raw_key": val_raw_keys.values,
    "engineered_key": val_engineered_keys.values,
    "label": y_val.values
})


# Current model predictions
raw_analysis["probability"] = val_prob
raw_analysis["prediction"] = val_predictions


# Focus on benign false positives
benign2_fp = raw_analysis[
    (raw_analysis["label"] == 0)
    &
    (raw_analysis["prediction"] == 1)
].copy()


# ============================================================
# 7. Match benign_2 FPs against COMPLETE raw training pattern
# ============================================================

benign2_fp = benign2_fp.merge(
    train_raw_stats,
    left_on="raw_key",
    right_on="key",
    how="left"
)

benign2_fp["raw_pattern_seen_in_train"] = (
    benign2_fp["train_count"].notna()
)

benign2_fp["raw_pattern_seen_as_attack"] = (
    benign2_fp["train_attack_count"]
    .fillna(0) > 0
)


# ============================================================
# 8. Compare engineered vs raw pattern overlap
# ============================================================

print("\n" + "=" * 80)
print("ENGINEERED VS RAW PATTERN OVERLAP")
print("=" * 80)

total_fp = len(benign2_fp)

engineered_seen = false_positives[
    "seen_in_train"
].sum()

engineered_attack = false_positives[
    "seen_as_attack_in_train"
].sum()

raw_seen = benign2_fp[
    "raw_pattern_seen_in_train"
].sum()

raw_attack = benign2_fp[
    "raw_pattern_seen_as_attack"
].sum()

print(
    f"Total benign_2 false positives: {total_fp}"
)

print("\nENGINEERED 19 FEATURES")

print(
    "Pattern seen in training:",
    engineered_seen,
    f"({engineered_seen / total_fp * 100:.2f}%)"
)

print(
    "Pattern seen as attack:",
    engineered_attack,
    f"({engineered_attack / total_fp * 100:.2f}%)"
)

print("\nCOMPLETE ORIGINAL 27 FEATURES")

print(
    "Exact raw pattern seen in training:",
    raw_seen,
    f"({raw_seen / total_fp * 100:.2f}%)"
)

print(
    "Exact raw pattern seen as attack:",
    raw_attack,
    f"({raw_attack / total_fp * 100:.2f}%)"
)


# ============================================================
# 9. Information preserved vs lost
# ============================================================

print("\n" + "=" * 80)
print("INFORMATION-LOSS INDICATOR")
print("=" * 80)

engineered_only = (
    (false_positives["seen_in_train"])
)

# Raw key lookup for every benign_2 FP
raw_lookup = pd.DataFrame({
    "raw_key": val_raw_keys.values
})

raw_train_key_set = set(
    train_raw_stats["key"]
)

fp_raw_seen = raw_lookup.loc[
    benign2_fp.index,
    "raw_key"
].isin(raw_train_key_set)
```

```python
import pandas as pd
import numpy as np

print("=" * 80)
print("STATEFUL FEATURE-PATTERN AMBIGUITY ANALYSIS")
print("=" * 80)


# ============================================================
# 1. Build training feature-pattern statistics
# ============================================================

train_pattern = pd.DataFrame({
    "feature_key": make_feature_key(X_train).values,
    "label": y_train.values
})

pattern_stats = (
    train_pattern
    .groupby("feature_key")["label"]
    .agg(
        total_rows="size",
        benign_rows=lambda x: (x == 0).sum(),
        attack_rows=lambda x: (x == 1).sum()
    )
    .reset_index()
)

pattern_stats["attack_rate"] = (
    pattern_stats["attack_rows"]
    / pattern_stats["total_rows"]
)

pattern_stats["has_both_labels"] = (
    (pattern_stats["benign_rows"] > 0)
    &
    (pattern_stats["attack_rows"] > 0)
)


# ============================================================
# 2. Overall training ambiguity
# ============================================================

total_training_rows = len(train_pattern)

conflicting_patterns = pattern_stats[
    pattern_stats["has_both_labels"]
]

conflicting_pattern_count = len(
    conflicting_patterns
)

# Rows belonging to conflicting patterns
conflicting_pattern_keys = set(
    conflicting_patterns["feature_key"]
)

conflicting_rows = train_pattern[
    train_pattern["feature_key"].isin(
        conflicting_pattern_keys
    )
]

conflicting_row_count = len(
    conflicting_rows
)


print("\n" + "=" * 80)
print("TRAINING PATTERN STATISTICS")
print("=" * 80)

print(
    "Total unique feature patterns:",
    len(pattern_stats)
)

print(
    "Patterns containing BOTH labels:",
    conflicting_pattern_count
)

print(
    "Percentage of patterns containing both labels:",
    round(
        conflicting_pattern_count
        / len(pattern_stats) * 100,
        2
    ),
    "%"
)

print(
    "\nRows belonging to conflicting patterns:",
    conflicting_row_count
)

print(
    "Percentage of training rows belonging to conflicting patterns:",
    round(
        conflicting_row_count
        / total_training_rows * 100,
        2
    ),
    "%"
)


# ============================================================
# 3. Majority-label theoretical accuracy
# ============================================================

# For each exact feature pattern, predict whichever label
# occurs most frequently in training.

pattern_stats["majority_count"] = (
    pattern_stats[
        ["benign_rows", "attack_rows"]
    ].max(axis=1)
)

majority_correct = (
    pattern_stats["majority_count"].sum()
)

majority_accuracy = (
    majority_correct
    / total_training_rows
)


print("\n" + "=" * 80)
print("MAJORITY-PATTERN CLASSIFIER")
print("=" * 80)

print(
    "Training accuracy if each feature pattern is assigned "
    "its majority class:",
    round(majority_accuracy, 4)
)

print(
    "Equivalent percentage:",
    round(majority_accuracy * 100, 2),
    "%"
)

print(
    "\nThis is not a model benchmark."
)

print(
    "It measures how much deterministic information the "
    "feature representation contains at exact-pattern level."
)


# ============================================================
# 4. Pattern ambiguity severity
# ============================================================

print("\n" + "=" * 80)
print("CONFLICTING PATTERN SEVERITY")
print("=" * 80)

conflict_distribution = (
    conflicting_patterns["attack_rate"]
    .describe()
)

print(
    conflict_distribution.round(4).to_string()
)


# ============================================================
# 5. Categorize training patterns
# ============================================================

def classify_pattern(rate):

    if rate == 0:
        return "Benign-only"

    elif rate == 1:
        return "Attack-only"

    else:
        return "Mixed"


pattern_stats["pattern_type"] = (
    pattern_stats["attack_rate"]
    .apply(classify_pattern)
)

pattern_type_summary = (
    pattern_stats
    .groupby("pattern_type")
    .agg(
        unique_patterns=("feature_key", "size"),
        rows=("total_rows", "sum")
    )
    .reset_index()
)

pattern_type_summary["pattern_percentage"] = (
    pattern_type_summary["unique_patterns"]
    / len(pattern_stats)
    * 100
)

pattern_type_summary["row_percentage"] = (
    pattern_type_summary["rows"]
    / total_training_rows
    * 100
)

print(
    pattern_type_summary.to_string(
        index=False,
        formatters={
            "pattern_percentage":
                "{:.2f}%".format,
            "row_percentage":
                "{:.2f}%".format
        }
    )
)


# ============================================================
# 6. Analyze benign_2 validation patterns
# ============================================================

val_pattern = pd.DataFrame({
    "feature_key": make_feature_key(X_val).values,
    "label": y_val.values,
    "probability": val_prob,
    "prediction": val_predictions
})

benign2_pattern = val_pattern[
    val_pattern["label"] == 0
].copy()

benign2_pattern = benign2_pattern.merge(
    pattern_stats[
        [
            "feature_key",
            "total_rows",
            "benign_rows",
            "attack_rows",
            "attack_rate",
            "pattern_type"
        ]
    ],
    on="feature_key",
    how="left"
)

# Patterns not seen in training
benign2_pattern["pattern_type"] = (
    benign2_pattern["pattern_type"]
    .fillna("Unseen")
)


# ============================================================
# 7. Benign_2 pattern categories
# ============================================================

print("\n" + "=" * 80)
print("BENIGN_2 PATTERN CATEGORIES")
print("=" * 80)

benign2_category = (
    benign2_pattern
    .groupby("pattern_type")
    .agg(
        rows=("feature_key", "size"),
        average_probability=("probability", "mean")
    )
    .reset_index()
)

benign2_category["percentage"] = (
    benign2_category["rows"]
    / len(benign2_pattern)
    * 100
)

print(
    benign2_category.to_string(
        index=False,
        formatters={
            "average_probability":
                "{:.4f}".format,
            "percentage":
                "{:.2f}%".format
        }
    )
)


# ============================================================
# 8. Attack-dominant patterns inside benign_2
# ============================================================

attack_dominant = benign2_pattern[
    benign2_pattern["attack_rate"] > 0.5
]

benign_dominant = benign2_pattern[
    benign2_pattern["attack_rate"] < 0.5
]

mixed_patterns = benign2_pattern[
    (benign2_pattern["attack_rate"] > 0)
    &
    (benign2_pattern["attack_rate"] < 1)
]


print("\n" + "=" * 80)
print("BENIGN_2 LABEL AMBIGUITY")
print("=" * 80)

print(
    "Benign_2 rows with attack-dominant training patterns:",
    len(attack_dominant),
    f"({len(attack_dominant) / len(benign2_pattern) * 100:.2f}%)"
)

print(
    "Benign_2 rows with benign-dominant training patterns:",
    len(benign_dominant),
    f"({len(benign_dominant) / len(benign2_pattern) * 100:.2f}%)"
)

print(
    "Benign_2 rows with mixed-label training patterns:",
    len(mixed_patterns),
    f"({len(mixed_patterns) / len(benign2_pattern) * 100:.2f}%)"
)


# ============================================================
# 9. Most ambiguous training patterns
# ============================================================

print("\n" + "=" * 80)
print("MOST AMBIGUOUS TRAINING PATTERNS")
print("=" * 80)

# Distance from 0.5 = lower means more ambiguous
ambiguous_patterns = (
    conflicting_patterns
    .copy()
)

ambiguous_patterns["ambiguity_score"] = (
    0.5
    - (
        ambiguous_patterns["attack_rate"]
        - 0.5
    ).abs()
)

ambiguous_patterns = (
    ambiguous_patterns
    .sort_values(
        "ambiguity_score",
        ascending=False
    )
)

print(
    ambiguous_patterns[
        [
            "total_rows",
            "benign_rows",
            "attack_rows",
            "attack_rate",
            "ambiguity_score"
        ]
    ]
    .head(25)
    .to_string(
        index=False,
        formatters={
            "attack_rate":
                "{:.4f}".format,
            "ambiguity_score":
                "{:.4f}".format
        }
    )
)


# ============================================================
# 10. Final diagnosis
# ============================================================

print("\n" + "=" * 80)
print("FINAL DIAGNOSIS")
print("=" * 80)

print(
    "Training rows:",
    total_training_rows
)

print(
    "Conflicting-pattern rows:",
    conflicting_row_count
)

print(
    "Conflicting-row percentage:",
    round(
        conflicting_row_count
        / total_training_rows * 100,
        2
    ),
    "%"
)

print(
    "Majority-pattern training accuracy:",
    round(
        majority_accuracy * 100,
        2
    ),
    "%"
)

print("\n✓ Ambiguity analysis complete.")
print("✓ Validation and training data only.")
print("✓ Test data was NOT used.")
```

```python
import pandas as pd
import numpy as np

print("=" * 80)
print("STATEFUL CONTEXT FEATURE ENGINEERING - CORRECTED")
print("=" * 80)


# ============================================================
# 1. Start from the ALREADY ENGINEERED numeric features
# ============================================================

# df_features was created earlier from df_stateful.
# It contains the engineered numeric columns such as:
# country_count, ttl_unique_count, etc.

df_context = df_features.copy()

# Add source ONLY for grouping.
# source is never used as a model feature.
df_context["source"] = (
    df_stateful["source"]
    .reindex(df_context.index)
)

# Preserve the original row order.
df_context["_original_order"] = np.arange(
    len(df_context)
)


print("Rows:", len(df_context))
print("Sources:", df_context["source"].nunique())

print(
    "Engineered base columns available:",
    len(
        df_context.drop(
            columns=[
                "label",
                "source",
                "_original_order"
            ]
        ).columns
    )
)


# ============================================================
# 2. Sort within each capture
# ============================================================

df_context = df_context.sort_values(
    ["source", "_original_order"]
).copy()


# ============================================================
# 3. Base features for temporal context
# ============================================================

context_features = [
    "rr",
    "PTR_frequency",
    "rr_count",
    "rr_name_entropy",
    "rr_name_length",
    "distinct_ns",
    "ttl_mean",
    "ttl_variance",
    "country_count",
    "asn_count",
    "domain_count",
    "ttl_unique_count",
    "ttl_range",
    "reverse_dns_known",
    "is_A_record",
    "is_PTR_record",
    "rr_entropy_per_length",
    "ttl_variance_ratio"
]

# Verify all columns exist
missing_context_features = [
    feature
    for feature in context_features
    if feature not in df_context.columns
]

if missing_context_features:

    raise ValueError(
        "Missing context features: "
        + str(missing_context_features)
    )


print("\nContext features:")
for feature in context_features:
    print("  -", feature)


# ============================================================
# 4. Create causal rolling features
# ============================================================

print("\n" + "=" * 80)
print("CREATING CAUSAL ROLLING FEATURES")
print("=" * 80)

# We process each capture separately.
#
# For a row at position i:
#
#   previous_5_mean
#       = mean of rows i-5 ... i-1
#
#   previous_20_mean
#       = mean of rows i-20 ... i-1
#
# Therefore current/future rows are never used.

for source_name, source_indices in (
    df_context.groupby("source", sort=False).groups.items()
):

    source_data = df_context.loc[
        source_indices
    ].copy()

    print(
        f"Processing {source_name}: "
        f"{len(source_data)} rows"
    )

    for feature in context_features:

        series = source_data[feature]

        # Previous row only
        shifted = series.shift(1)

        # --------------------------------------------
        # Previous 5 records
        # --------------------------------------------

        mean_5 = (
            shifted
            .rolling(
                window=5,
                min_periods=1
            )
            .mean()
        )

        std_5 = (
            shifted
            .rolling(
                window=5,
                min_periods=2
            )
            .std()
        )

        # --------------------------------------------
        # Previous 20 records
        # --------------------------------------------

        mean_20 = (
            shifted
            .rolling(
                window=20,
                min_periods=1
            )
            .mean()
        )

        std_20 = (
            shifted
            .rolling(
                window=20,
                min_periods=2
            )
            .std()
        )

        df_context.loc[
            source_indices,
            f"{feature}_prev5_mean"
        ] = mean_5.values

        df_context.loc[
            source_indices,
            f"{feature}_prev5_std"
        ] = std_5.values

        df_context.loc[
            source_indices,
            f"{feature}_prev20_mean"
        ] = mean_20.values

        df_context.loc[
            source_indices,
            f"{feature}_prev20_std"
        ] = std_20.values


# ============================================================
# 5. Previous-record delta features
# ============================================================

print("\n" + "=" * 80)
print("CREATING PREVIOUS-RECORD DELTA FEATURES")
print("=" * 80)

delta_features = [
    "rr",
    "PTR_frequency",
    "rr_name_entropy",
    "rr_name_length",
    "ttl_mean",
    "ttl_variance"
]

for source_name, source_indices in (
    df_context.groupby("source", sort=False).groups.items()
):

    source_data = df_context.loc[
        source_indices
    ].copy()

    for feature in delta_features:

        previous = source_data[
            feature
        ].shift(1)

        delta = (
            source_data[feature]
            - previous
        )

        df_context.loc[
            source_indices,
            f"{feature}_delta_prev"
        ] = delta.values


# ============================================================
# 6. Deviation from previous 20-record context
# ============================================================

print("\n" + "=" * 80)
print("CREATING CONTEXT-DEVIATION FEATURES")
print("=" * 80)

deviation_features = [
    "rr",
    "PTR_frequency",
    "rr_name_entropy",
    "rr_name_length",
    "ttl_mean",
    "ttl_variance"
]

for source_name, source_indices in (
    df_context.groupby("source", sort=False).groups.items()
):

    source_data = df_context.loc[
        source_indices
    ].copy()

    for feature in deviation_features:

        previous = source_data[
            feature
        ].shift(1)

        previous_mean = (
            previous
            .rolling(
                window=20,
                min_periods=1
            )
            .mean()
        )

        deviation = (
            source_data[feature]
            - previous_mean
        )

        df_context.loc[
            source_indices,
            f"{feature}_vs_prev20"
        ] = deviation.values


# ============================================================
# 7. Previous-20 PTR / RR ratio
# ============================================================

print("\n" + "=" * 80)
print("CREATING PREVIOUS-20 PTR/RR RATIO")
print("=" * 80)

for source_name, source_indices in (
    df_context.groupby("source", sort=False).groups.items()
):

    source_data = df_context.loc[
        source_indices
    ].copy()

    previous_rr = (
        source_data["rr"]
        .shift(1)
        .rolling(
            window=20,
            min_periods=1
        )
        .mean()
    )

    previous_ptr = (
        source_data["PTR_frequency"]
        .shift(1)
        .rolling(
            window=20,
            min_periods=1
        )
        .mean()
    )

    ratio = (
        previous_ptr
        /
        previous_rr.clip(
            lower=1e-6
        )
    )

    df_context.loc[
        source_indices,
        "prev20_ptr_rr_ratio"
    ] = ratio.values


# ============================================================
# 8. Restore original row order
# ============================================================

df_context = df_context.sort_values(
    "_original_order"
).copy()


# ============================================================
# 9. Check new features
# ============================================================

helper_columns = [
    "label",
    "source",
    "_original_order"
]

base_columns = set(
    df_features.columns
)

new_context_features = [
    column
    for column in df_context.columns
    if column not in base_columns
    and column not in helper_columns
]

print("\n" + "=" * 80)
print("NEW CONTEXT FEATURES")
print("=" * 80)

print(
    "Number of new context features:",
    len(new_context_features)
)

for i, feature in enumerate(
    new_context_features,
    1
):
    print(f"{i:2d}. {feature}")


# ============================================================
# 10. Handle NaN / infinity
# ============================================================

print("\n" + "=" * 80)
print("MISSING / INFINITE VALUES")
print("=" * 80)

before_missing = (
    df_context[
        new_context_features
    ].isna().sum().sum()
)

before_inf = (
    np.isinf(
        df_context[
            new_context_features
        ].to_numpy()
    ).sum()
)

print(
    "Missing values before filling:",
    before_missing
)

print(
    "Infinite values before filling:",
    before_inf
)


df_context[
    new_context_features
] = (
    df_context[
        new_context_features
    ]
    .replace(
        [np.inf, -np.inf],
        np.nan
    )
    .fillna(0)
)


# ============================================================
# 11. Final validation
# ============================================================

after_missing = (
    df_context[
        new_context_features
    ].isna().sum().sum()
)

after_inf = (
    np.isinf(
        df_context[
            new_context_features
        ].to_numpy()
    ).sum()
)

print(
    "Missing values after filling:",
    after_missing
)

print(
    "Infinite values after filling:",
    after_inf
)


# ============================================================
# 12. Verify source boundaries
# ============================================================

print("\n" + "=" * 80)
print("SOURCE BOUNDARY CHECK")
print("=" * 80)

source_sizes = (
    df_context
    .groupby("source")
    .size()
)

print(source_sizes)


# ============================================================
# 13. Check that row order was preserved
# ============================================================

print("\n" + "=" * 80)
print("ROW ORDER CHECK")
print("=" * 80)

original_indices = (
    df_stateful.index
)

context_indices = (
    df_context.index
)

print(
    "Same number of rows:",
    len(original_indices) == len(context_indices)
)

print(
    "Same row index set:",
    set(original_indices) == set(context_indices)
)

print(
    "First 5 original indices:",
    list(original_indices[:5])
)

print(
    "First 5 context indices:",
    list(context_indices[:5])
)


# ============================================================
# 14. Context feature statistics
# ============================================================

print("\n" + "=" * 80)
print("CONTEXT FEATURE STATISTICS")
print("=" * 80)

print(
    df_context[
        new_context_features
    ].describe()
    .round(4)
    .to_string()
)


# ============================================================
# 15. Compare context means by class
# ============================================================

print("\n" + "=" * 80)
print("TOP CONTEXT FEATURES BY CLASS DIFFERENCE")
print("=" * 80)

context_class_means = pd.DataFrame({

    "Benign_Mean":
        df_context[
            df_context["label"] == 0
        ][new_context_features].mean(),

    "Attack_Mean":
        df_context[
            df_context["label"] == 1
        ][new_context_features].mean()
})

context_class_means[
    "Absolute_Difference"
] = (
    context_class_means["Attack_Mean"]
    - context_class_means["Benign_Mean"]
).abs()

context_class_means = (
    context_class_means
    .sort_values(
        "Absolute_Difference",
        ascending=False
    )
)

print(
    context_class_means
    .head(30)
    .round(4)
    .to_string()
)


# ============================================================
# 16. Save final contextual dataset
# ============================================================

df_stateful_context = df_context.copy()

# Remove helper columns from the saved dataset
df_stateful_context = (
    df_stateful_context
    .drop(
        columns=["_original_order"]
    )
)


print("\n" + "=" * 80)
print("✓ CONTEXT FEATURE ENGINEERING COMPLETE")
print("=" * 80)

print(
    "Original engineered columns:",
    len(df_features.columns)
)

print(
    "New context features:",
    len(new_context_features)
)

print(
    "Final columns:",
    len(df_stateful_context.columns)
)

print(
    "Final rows:",
    len(df_stateful_context)
)

print(
    "Missing values:",
    df_stateful_context.isna().sum().sum()
)

print(
    "Test labels were not used."
)
```

```python
import numpy as np
import pandas as pd
import xgboost as xgb

from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("=" * 80)
print("OOF THRESHOLD CALIBRATION - CORRECTED")
print("=" * 80)


# ============================================================
# 1. DEVELOPMENT DATA ONLY
# ============================================================

X_dev = pd.concat(
    [
        X_train_context,
        X_val_context
    ],
    axis=0
)

y_dev = pd.concat(
    [
        y_train,
        y_val
    ],
    axis=0
)

groups_dev = groups.loc[
    X_dev.index
].copy()


print("Development rows:", len(X_dev))
print("Development features:", X_dev.shape[1])

print("\nClass distribution:")
print(
    y_dev.value_counts()
    .sort_index()
)

print("\nDevelopment sources:")
for source in sorted(groups_dev.unique()):
    print("  -", source)


# ============================================================
# 2. FIXED MODEL CONFIGURATION
# ============================================================

OOF_WEIGHT = 3.0
OOF_ESTIMATORS = 73


def create_oof_model():

    return xgb.XGBClassifier(

        n_estimators=OOF_ESTIMATORS,

        learning_rate=0.04,

        max_depth=5,

        min_child_weight=3,

        subsample=0.90,

        colsample_bytree=0.90,

        gamma=0.1,

        reg_alpha=0.1,

        reg_lambda=3.0,

        scale_pos_weight=OOF_WEIGHT,

        objective="binary:logistic",

        eval_metric="aucpr",

        random_state=42,

        n_jobs=-1,

        tree_method="hist"
    )


# ============================================================
# 3. GROUP-AWARE OOF SPLIT
# ============================================================

cv = StratifiedGroupKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

oof_prob = np.full(
    len(X_dev),
    np.nan,
    dtype=float
)


# ============================================================
# 4. GENERATE OOF PREDICTIONS
# ============================================================

print("\n" + "=" * 80)
print("GENERATING OOF PREDICTIONS")
print("=" * 80)

fold_records = []

for fold_number, (
    train_relative,
    valid_relative
) in enumerate(
    cv.split(
        X_dev,
        y_dev,
        groups_dev
    ),
    start=1
):

    X_fold_train = X_dev.iloc[
        train_relative
    ]

    y_fold_train = y_dev.iloc[
        train_relative
    ]

    X_fold_valid = X_dev.iloc[
        valid_relative
    ]

    y_fold_valid = y_dev.iloc[
        valid_relative
    ]

    train_sources = set(
        groups_dev.iloc[
            train_relative
        ].unique()
    )

    valid_sources = set(
        groups_dev.iloc[
            valid_relative
        ].unique()
    )

    print("\n" + "-" * 80)
    print(f"FOLD {fold_number}")
    print("-" * 80)

    print("Train rows:", len(train_relative))
    print("Validation rows:", len(valid_relative))

    print("Train sources:")
    for source in sorted(train_sources):
        print("  -", source)

    print("Validation sources:")
    for source in sorted(valid_sources):
        print("  -", source)

    print(
        "Validation classes:",
        y_fold_valid.value_counts()
        .sort_index()
        .to_dict()
    )

    # --------------------------------------------------------
    # Leakage check
    # --------------------------------------------------------

    assert train_sources.isdisjoint(
        valid_sources
    )

    # Training fold must contain both classes
    assert set(
        y_fold_train.unique()
    ) == {0, 1}

    # NOTE:
    # Validation fold is allowed to contain only one class.
    # This happens because each source contains only one class.
    # We only need the POOLED OOF predictions to contain both.
    # --------------------------------------------------------

    fold_model = create_oof_model()

    fold_model.fit(
        X_fold_train,
        y_fold_train,
        verbose=False
    )

    fold_probabilities = fold_model.predict_proba(
        X_fold_valid
    )[:, 1]

    oof_prob[
        valid_relative
    ] = fold_probabilities

    fold_records.append({
        "fold": fold_number,
        "train_rows": len(train_relative),
        "validation_rows": len(valid_relative),
        "train_sources": sorted(train_sources),
        "validation_sources": sorted(valid_sources),
        "validation_benign": int(
            (y_fold_valid == 0).sum()
        ),
        "validation_attack": int(
            (y_fold_valid == 1).sum()
        )
    })


# ============================================================
# 5. VERIFY OOF COVERAGE
# ============================================================

print("\n" + "=" * 80)
print("OOF COVERAGE")
print("=" * 80)

missing_oof = np.isnan(oof_prob).sum()

print(
    "Total development rows:",
    len(oof_prob)
)

print(
    "Rows with OOF prediction:",
    np.sum(~np.isnan(oof_prob))
)

print(
    "Rows without OOF prediction:",
    missing_oof
)

assert missing_oof == 0


# ============================================================
# 6. POOLED OOF CLASS CHECK
# ============================================================

print("\n" + "=" * 80)
print("POOLED OOF CLASS CHECK")
print("=" * 80)

print(
    "Benign rows:",
    int((y_dev == 0).sum())
)

print(
    "Attack rows:",
    int((y_dev == 1).sum())
)

assert set(
    y_dev.unique()
) == {0, 1}


# ============================================================
# 7. THRESHOLD SEARCH
# ============================================================

print("\n" + "=" * 80)
print("POOLED OOF THRESHOLD SEARCH")
print("=" * 80)

thresholds = np.arange(
    0.001,
    1.000,
    0.001
)

threshold_results = []

for threshold in thresholds:

    predictions = (
        oof_prob >= threshold
    ).astype(int)

    precision = precision_score(
        y_dev,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_dev,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_dev,
        predictions,
        zero_division=0
    )

    tn, fp, fn, tp = confusion_matrix(
        y_dev,
        predictions
    ).ravel()

    fpr = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else 0
    )

    threshold_results.append({
        "threshold": threshold,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "fpr": fpr,
        "tn": tn,
        "fp": fp,
        "fn": fn,
        "tp": tp
    })


oof_threshold_df = pd.DataFrame(
    threshold_results
)


# ============================================================
# 8. BEST F1 THRESHOLD
# ============================================================

best_f1_row = oof_threshold_df.loc[
    oof_threshold_df["f1"].idxmax()
]

best_oof_threshold = float(
    best_f1_row["threshold"]
)


print("\n" + "=" * 80)
print("BEST OOF F1 THRESHOLD")
print("=" * 80)

print(
    "Threshold:",
    round(best_oof_threshold, 4)
)

print(
    "Precision:",
    round(best_f1_row["precision"], 4)
)

print(
    "Recall:",
    round(best_f1_row["recall"], 4)
)

print(
    "F1:",
    round(best_f1_row["f1"], 4)
)

print(
    "FPR:",
    round(best_f1_row["fpr"], 4)
)


# ============================================================
# 9. PRECISION-TARGET ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("PRECISION TARGET ANALYSIS")
print("=" * 80)

precision_targets = [
    0.70,
    0.75,
    0.80,
    0.85,
    0.90
]

target_results = []

for target in precision_targets:

    eligible = oof_threshold_df[
        oof_threshold_df["precision"] >= target
    ]

    if len(eligible) == 0:

        target_results.append({
            "target_precision": target,
            "threshold": np.nan,
            "precision": np.nan,
            "recall": np.nan,
            "f1": np.nan,
            "fpr": np.nan
        })

    else:

        # Among thresholds satisfying the target,
        # choose highest F1.
        row = eligible.loc[
            eligible["f1"].idxmax()
        ]

        target_results.append({
            "target_precision": target,
            "threshold": row["threshold"],
            "precision": row["precision"],
            "recall": row["recall"],
            "f1": row["f1"],
            "fpr": row["fpr"]
        })


target_df = pd.DataFrame(
    target_results
)

print(
    target_df.to_string(
        index=False,
        formatters={
            "target_precision":
                "{:.2f}".format,
            "threshold":
                lambda x:
                "N/A"
                if pd.isna(x)
                else f"{x:.4f}",
            "precision":
                lambda x:
                "N/A"
                if pd.isna(x)
                else f"{x:.4f}",
            "recall":
                lambda x:
                "N/A"
                if pd.isna(x)
                else f"{x:.4f}",
            "f1":
                lambda x:
                "N/A"
                if pd.isna(x)
                else f"{x:.4f}",
            "fpr":
                lambda x:
                "N/A"
                if pd.isna(x)
                else f"{x:.4f}"
        }
    )
)


# ============================================================
# 10. OLD THRESHOLD COMPARISON
# ============================================================

old_threshold = 0.02

old_row = oof_threshold_df.iloc[
    (
        oof_threshold_df["threshold"]
        - old_threshold
    ).abs().argmin()
]


print("\n" + "=" * 80)
print("OLD THRESHOLD VS OOF THRESHOLD")
print("=" * 80)

comparison = pd.DataFrame({

    "Threshold_0.02": [
        old_row["precision"],
        old_row["recall"],
        old_row["f1"],
        old_row["fpr"]
    ],

    "OOF_Best_F1": [
        best_f1_row["precision"],
        best_f1_row["recall"],
        best_f1_row["f1"],
        best_f1_row["fpr"]
    ]

}, index=[
    "Precision",
    "Recall",
    "F1",
    "FPR"
])

print(
    comparison.round(4).to_string()
)


# ============================================================
# 11. LOCAL REGION AROUND BEST THRESHOLD
# ============================================================

print("\n" + "=" * 80)
print("THRESHOLDS AROUND BEST OOF THRESHOLD")
print("=" * 80)

near_best = oof_threshold_df[
    (
        oof_threshold_df["threshold"]
        >= best_oof_threshold - 0.05
    )
    &
    (
        oof_threshold_df["threshold"]
        <= best_oof_threshold + 0.05
    )
]

# Print every 5th threshold to keep output readable.
near_best = near_best.iloc[::5]

print(
    near_best[
        [
            "threshold",
            "precision",
            "recall",
            "f1",
            "fpr"
        ]
    ].to_string(
        index=False,
        formatters={
            "threshold":
                "{:.3f}".format,
            "precision":
                "{:.4f}".format,
            "recall":
                "{:.4f}".format,
            "f1":
                "{:.4f}".format,
            "fpr":
                "{:.4f}".format
        }
    )
)


# ============================================================
# 12. SAVE CALIBRATION RESULTS
# ============================================================

oof_calibration = {
    "threshold": best_oof_threshold,
    "weight": OOF_WEIGHT,
    "estimators": OOF_ESTIMATORS
}


# ============================================================
# 13. FINAL STATUS
# ============================================================

print("\n" + "=" * 80)
print("✓ OOF CALIBRATION COMPLETE")
print("=" * 80)

print(
    "Recommended threshold:",
    round(best_oof_threshold, 4)
)

print(
    "Test data used: NO"
)

print(
    "Test labels used for threshold selection: NO"
)
```

```python
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

print("=" * 80)
print("FINAL TEST EVALUATION — OOF CALIBRATED THRESHOLD")
print("=" * 80)


# ============================================================
# 1. Use threshold selected ONLY from OOF development data
# ============================================================

FINAL_CALIBRATED_THRESHOLD = best_oof_threshold

print(
    "OOF-selected threshold:",
    round(FINAL_CALIBRATED_THRESHOLD, 4)
)

assert (
    FINAL_CALIBRATED_THRESHOLD
    == best_oof_threshold
)


# ============================================================
# 2. Use the FINAL MODEL trained on TRAIN + VALIDATION
# ============================================================

# final_test_prob was already generated by final_model
# in Code 23.

final_prob = final_test_prob.copy()

final_pred_calibrated = (
    final_prob >= FINAL_CALIBRATED_THRESHOLD
).astype(int)


# ============================================================
# 3. Calculate metrics
# ============================================================

tn, fp, fn, tp = confusion_matrix(
    y_test,
    final_pred_calibrated
).ravel()

accuracy = accuracy_score(
    y_test,
    final_pred_calibrated
)

precision = precision_score(
    y_test,
    final_pred_calibrated,
    zero_division=0
)

recall = recall_score(
    y_test,
    final_pred_calibrated,
    zero_division=0
)

f1 = f1_score(
    y_test,
    final_pred_calibrated,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_test,
    final_prob
)

pr_auc = average_precision_score(
    y_test,
    final_prob
)

fpr = (
    fp / (fp + tn)
    if (fp + tn) > 0
    else 0
)

specificity = (
    tn / (tn + fp)
    if (tn + fp) > 0
    else 0
)


# ============================================================
# 4. Final metrics
# ============================================================

print("\n" + "=" * 80)
print("FINAL TEST METRICS")
print("=" * 80)

print(
    f"Accuracy    : {accuracy:.4f} "
    f"({accuracy * 100:.2f}%)"
)

print(
    f"Precision   : {precision:.4f} "
    f"({precision * 100:.2f}%)"
)

print(
    f"Recall      : {recall:.4f} "
    f"({recall * 100:.2f}%)"
)

print(
    f"F1          : {f1:.4f} "
    f"({f1 * 100:.2f}%)"
)

print(
    f"ROC-AUC     : {roc_auc:.4f} "
    f"({roc_auc * 100:.2f}%)"
)

print(
    f"PR-AUC      : {pr_auc:.4f} "
    f"({pr_auc * 100:.2f}%)"
)

print(
    f"FPR         : {fpr:.4f} "
    f"({fpr * 100:.2f}%)"
)

print(
    f"Specificity : {specificity:.4f} "
    f"({specificity * 100:.2f}%)"
)


# ============================================================
# 5. Confusion matrix
# ============================================================

print("\n" + "=" * 80)
print("FINAL CONFUSION MATRIX")
print("=" * 80)

confusion_df = pd.DataFrame(
    [
        [tn, fp],
        [fn, tp]
    ],
    index=[
        "Actual Benign",
        "Actual Attack"
    ],
    columns=[
        "Predicted Benign",
        "Predicted Attack"
    ]
)

print(confusion_df)


# ============================================================
# 6. Classification report
# ============================================================

print("\n" + "=" * 80)
print("CLASSIFICATION REPORT")
print("=" * 80)

print(
    classification_report(
        y_test,
        final_pred_calibrated,
        target_names=[
            "Benign",
            "Attack"
        ],
        digits=4,
        zero_division=0
    )
)


# ============================================================
# 7. Compare old threshold vs calibrated threshold
#
# This is descriptive only.
# The calibrated threshold was selected without test data.
# ============================================================

old_pred = (
    final_prob >= 0.02
).astype(int)

old_tn, old_fp, old_fn, old_tp = confusion_matrix(
    y_test,
    old_pred
).ravel()

old_precision = precision_score(
    y_test,
    old_pred,
    zero_division=0
)

old_recall = recall_score(
    y_test,
    old_pred,
    zero_division=0
)

old_f1 = f1_score(
    y_test,
    old_pred,
    zero_division=0
)

old_fpr = (
    old_fp / (old_fp + old_tn)
    if (old_fp + old_tn) > 0
    else 0
)


comparison = pd.DataFrame({

    "Threshold_0.02": [
        old_precision,
        old_recall,
        old_f1,
        old_fpr,
        old_fp,
        old_fn
    ],

    "OOF_Threshold": [
        precision,
        recall,
        f1,
        fpr,
        fp,
        fn
    ]

}, index=[
    "Precision",
    "Recall",
    "F1",
    "FPR",
    "False Positives",
    "False Negatives"
])


print("\n" + "=" * 80)
print("OLD VS OOF-CALIBRATED THRESHOLD")
print("=" * 80)

print(
    comparison.round(4).to_string()
)


# ============================================================
# 8. Per-capture final results
# ============================================================

print("\n" + "=" * 80)
print("PER-CAPTURE TEST RESULTS")
print("=" * 80)

test_sources_final = (
    groups
    .loc[X_test.index]
    .reset_index(drop=True)
)

test_labels_final = (
    y_test
    .reset_index(drop=True)
)

test_predictions_final = pd.Series(
    final_pred_calibrated
)

test_probabilities_final = pd.Series(
    final_prob
)

capture_results = []

for source in sorted(
    test_sources_final.unique()
):

    mask = (
        test_sources_final == source
    )

    source_y = test_labels_final[mask]
    source_pred = test_predictions_final[mask]
    source_prob = test_probabilities_final[mask]

    tn_s, fp_s, fn_s, tp_s = confusion_matrix(
        source_y,
        source_pred,
        labels=[0, 1]
    ).ravel()

    precision_s = precision_score(
        source_y,
        source_pred,
        zero_division=0
    )

    recall_s = recall_score(
        source_y,
        source_pred,
        zero_division=0
    )

    f1_s = f1_score(
        source_y,
        source_pred,
        zero_division=0
    )

    fpr_s = (
        fp_s / (fp_s + tn_s)
        if (fp_s + tn_s) > 0
        else 0
    )

    capture_results.append({
        "source": source,
        "samples": len(source_y),
        "benign": int((source_y == 0).sum()),
        "attack": int((source_y == 1).sum()),
        "precision": precision_s,
        "recall": recall_s,
        "f1": f1_s,
        "fpr": fpr_s,
        "TN": tn_s,
        "FP": fp_s,
        "FN": fn_s,
        "TP": tp_s,
        "mean_probability": source_prob.mean(),
        "median_probability": source_prob.median()
    })


capture_df_final = pd.DataFrame(
    capture_results
)

print(
    capture_df_final.to_string(
        index=False,
        formatters={
            "precision": "{:.4f}".format,
            "recall": "{:.4f}".format,
            "f1": "{:.4f}".format,
            "fpr": "{:.4f}".format,
            "mean_probability": "{:.4f}".format,
            "median_probability": "{:.4f}".format
        }
    )
)


# ============================================================
# 9. Final test score distribution
# ============================================================

print("\n" + "=" * 80)
print("FINAL TEST SCORE DISTRIBUTION")
print("=" * 80)

for label, name in [
    (0, "Benign"),
    (1, "Attack")
]:

    values = final_prob[
        y_test.values == label
    ]

    print(
        f"{name}: "
        f"count={len(values)}, "
        f"min={np.min(values):.4f}, "
        f"P25={np.percentile(values, 25):.4f}, "
        f"median={np.median(values):.4f}, "
        f"P75={np.percentile(values, 75):.4f}, "
        f"P95={np.percentile(values, 95):.4f}, "
        f"max={np.max(values):.4f}"
    )


# ============================================================
# 10. Final integrity
# ============================================================

print("\n" + "=" * 80)
print("FINAL EXPERIMENT STATUS")
print("=" * 80)

print("Final training rows :", len(X_train_final))
print("Final test rows     :", len(X_test_context))
print("Final features      :", len(selected_context_features))

print(
    "Threshold source:",
    "OOF development predictions"
)

print(
    "Test data used for threshold selection:",
    "NO"
)

print(
    "Test data used during final training:",
    "NO"
)

print("\n✓ FINAL FROZEN TEST EVALUATION COMPLETE")
```
